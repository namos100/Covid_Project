```python
# scripts/01_preprocessing.ipynb

from pathlib import Path
import pandas as pd
import numpy as np

RAW_PATH = Path("/home/bcrlab/igguest/porat_naama/data/raw/geographic-sum-per-day-ver_00859.csv")
OUT_PATH = Path("/home/bcrlab/igguest/porat_naama/data/processed/covid_clean.parquet")

df_loc_corona = pd.read_csv(RAW_PATH).copy()
(df_loc_corona == '<15').any().any()


```

```python
df_loc_corona.replace('<15', 7, inplace=True)
print((df_loc_corona == '<15').any().any())
df_loc_corona = df_loc_corona.fillna(0)
df_loc_corona.head()
```

```python
df_loc_corona['agas_code'] = df_loc_corona['agas_code'].astype(int)
df_loc_corona['City_agas_code'] = df_loc_corona['town_code'].astype(str) + "_" + df_loc_corona['agas_code'].astype(str)
df_loc_corona = df_loc_corona.set_index('City_agas_code')

df_loc_corona.head()
```

```python
df_loc_corona = df_loc_corona.drop(columns=['new_cases_on_date', 'new_recoveries_on_date', 'new_hospitalized_on_date', 'new_deaths_on_date', 'new_diagnostic_tests_on_date', 'new_vacc_first_dose_on_date', 'new_vacc_second_dose_on_date', 'new_vacc_third_dose_on_date'])
df_loc_corona = df_loc_corona.drop(columns=['town_code','agas_code'])
df_loc_corona = df_loc_corona.sort_values(['City_agas_code', 'date'])
df_loc_corona.head()
```

```python
columns_to_convert = ['accumulated_cases', 'accumulated_recoveries', 'accumulated_hospitalized', 'accumulated_deaths', 'accumulated_diagnostic_tests', 'accumulated_vaccination_first_dose', 'accumulated_vaccination_second_dose', 'accumulated_vaccination_third_dose']
for column in columns_to_convert:
    # Convert to numeric first, then cast to integer
    df_loc_corona[column] = pd.to_numeric(df_loc_corona[column], errors='coerce').fillna(0).astype(int)
df_loc_corona.info()
```

```python

df_loc_corona = df_loc_corona.copy()

# (also important: sort before diff)
df_loc_corona = df_loc_corona.sort_values(["City_agas_code", "date"])

df_loc_corona["positive_daily"]      = df_loc_corona.groupby("City_agas_code")["accumulated_cases"].diff()
df_loc_corona["hospitalized_daily"]  = df_loc_corona.groupby("City_agas_code")["accumulated_hospitalized"].diff()
df_loc_corona["tested_daily"]        = df_loc_corona.groupby("City_agas_code")["accumulated_diagnostic_tests"].diff()

df_loc_corona["infection_rate"]   = df_loc_corona["positive_daily"] / df_loc_corona["tested_daily"]
df_loc_corona["recovery_ratio"]   = df_loc_corona["accumulated_recoveries"] / df_loc_corona["accumulated_cases"]
df_loc_corona["recovery_index"]   = df_loc_corona["accumulated_recoveries"] / df_loc_corona["accumulated_diagnostic_tests"]

# Replace inf/-inf -> NaN, then decide how to fill
df_loc_corona["infection_rate"] = df_loc_corona["infection_rate"].replace([np.inf, -np.inf], np.nan)

# Now choose fill policy
df_loc_corona = df_loc_corona.fillna(0)
df_loc_corona.tail()
```

```python
df_loc_corona['is_lockdown'] = 0
df_loc_corona['date']= pd.to_datetime(df_loc_corona['date'])
df_loc_corona.info()
```

```python
#lockdown dates:
#March 19, 2020 – April 24, 2020
#September 18, 2020 – October 18, 2020
#December 27, 2020 – February 6, 2021
#July 29, 2021 – November 27, 2021

lockdown_periods = [('2020-03-19', '2020-04-24'), ('2020-09-18', '2020-10-18'), ('2020-12-27', '2021-02-06'), ('2021-07-29', '2021-11-27')]

lockdown_periods = [(pd.Timestamp(start), pd.Timestamp(end)) for start, end in lockdown_periods]


for start_date, end_date in lockdown_periods:
    mask = (df_loc_corona['date'] >= start_date) & (df_loc_corona['date'] <= end_date)
    df_loc_corona.loc[mask, 'is_lockdown'] = 1

df_loc_corona.tail()
```

```python
import re
import numpy as np
import pandas as pd
from pathlib import Path

Mifkad_RAW_PATH = Path("/home/bcrlab/igguest/porat_naama/data/raw/mifkad 2022.csv")
```

```python
raw_mif = pd.read_csv(Mifkad_RAW_PATH).copy()

raw_mif.head()
```

```python

# Clean key parts
raw_mif["LocalityCode"] = raw_mif["LocalityCode"].astype(str).str.strip()
raw_mif["StatArea"] = raw_mif["StatArea"].fillna(0)
raw_mif["StatArea"] = pd.to_numeric(raw_mif["StatArea"], errors="coerce").astype("Int64")

# Build key
raw_mif["City_agas_code"] = raw_mif["LocalityCode"] + "_" + raw_mif["StatArea"].astype(str)
raw_mif = raw_mif[raw_mif["StatArea"].notna()].copy()  # drop rows without a StatArea

# DO NOT fillna(0) globally here.
# Keep StatAreaCmb in raw_mif for mapping later.

out_dir = Path("data/outputs")
out_dir.mkdir(parents=True, exist_ok=True)

raw_mif_csv = out_dir / "raw_mif_inspect.csv"
raw_mif.to_csv(raw_mif_csv, index=False)


raw_mif.head()
```

```python
# Expand rows where StatAreaCmb lists multiple '+'-separated numbers: one duplicate
# row per number that isn't the row's own StatArea, identical in every other column.
print("raw_mif row count before expansion:", len(raw_mif))

_new_rows = []
for _, row in raw_mif.iterrows():
    cmb = str(row["StatAreaCmb"])
    if "+" not in cmb:
        continue
    own = str(row["StatArea"])
    for num in cmb.split("+"):
        num = num.strip()
        if num != own:
            new_row = row.copy()
            new_row["City_agas_code"] = str(row["LocalityCode"]) + "_" + num
            _new_rows.append(new_row)

raw_mif_expanded_simple = pd.concat([raw_mif, pd.DataFrame(_new_rows)], ignore_index=True)

print("raw_mif row count after expansion:", len(raw_mif_expanded_simple))
assert len(raw_mif_expanded_simple) == 3618, len(raw_mif_expanded_simple)
assert raw_mif_expanded_simple["City_agas_code"].is_unique
print("OK: City_agas_code unique, row count matches expected 3618.")
```

```python
raw_mif_expanded_simple = raw_mif_expanded_simple.drop(columns=["LocalityCode", "StatArea", "StatAreaCmb"])
raw_mif_expanded_simple = raw_mif_expanded_simple.set_index("City_agas_code")
raw_mif_expanded_simple.head()
```

```python
df_corona = pd.read_csv('/home/bcrlab/igguest/porat_naama/data/processed/Corona_loc_processed.csv', dtype={'City_agas_code': str})
df_mifkad = pd.read_csv('/home/bcrlab/igguest/porat_naama/data/processed/Mifkad_processed.csv',      dtype={'City_agas_code': str})

assert df_mifkad['City_agas_code'].is_unique, 'Mifkad key is not unique - merge would explode rows'
n_before = len(df_corona)

df_corona_merged = df_corona.merge(
    df_mifkad.drop(columns=['LocNameHeb']),   # 'town' already exists in the corona file
    on='City_agas_code',
    how='left',
    validate='many_to_one'
)
assert len(df_corona_merged) == n_before, 'Row count changed - unexpected duplication'

matched = df_corona_merged['pop'].notna()
keys_all = df_corona['City_agas_code'].unique()
keys_unmatched = sorted(set(keys_all) - set(df_mifkad['City_agas_code']))

print(f'rows:            {n_before:,} -> {len(df_corona_merged):,}')
print(f'rows matched:    {matched.sum():,} ({100*matched.mean():.1f}%)')
print(f'keys matched:    {len(keys_all)-len(keys_unmatched)} / {len(keys_all)}')
print(f'keys unmatched:  {len(keys_unmatched)}')

df_corona_merged['has_census'] = matched.astype(int)

(df_corona_merged.loc[~matched, ['City_agas_code', 'town']]
    .drop_duplicates()
    .to_csv('/home/bcrlab/igguest/porat_naama/data/processed/unmatched_keys.csv', index=False, encoding='utf-8-sig'))
```

```python
# --- inspect the unmatched keys ---
unmatched = (df_corona_merged.loc[df_corona_merged['has_census'] == 0]
             .groupby(['City_agas_code', 'town'], as_index=False)
             .agg(n_rows=('date', 'size'),
                  first_date=('date', 'min'),
                  last_date=('date', 'max'),
                  total_positive=('positive_daily', 'sum')))

# split the key back into its parts
unmatched['LocalityCode'] = unmatched['City_agas_code'].str.split('_').str[0]
unmatched['StatArea']     = unmatched['City_agas_code'].str.split('_').str[1]

# failure type: is the city missing from the census, or just this statistical area?
census_localities = set(df_mifkad['City_agas_code'].str.split('_').str[0])
unmatched['fail_type'] = unmatched['LocalityCode'].map(
    lambda l: 'locality_missing' if l not in census_localities else 'area_missing')

print(f'unmatched keys: {len(unmatched)}')
print(unmatched['fail_type'].value_counts(), '\n')

# which cities lose the most - these are the ones worth chasing
by_city = (unmatched.groupby(['LocalityCode', 'town', 'fail_type'], as_index=False)
           .agg(n_areas_lost=('City_agas_code', 'size'), n_rows=('n_rows', 'sum'))
           .sort_values('n_rows', ascending=False))
print(by_city.head(20).to_string(index=False))

unmatched.sort_values(['LocalityCode', 'StatArea']).to_csv(
    'unmatched_keys_detail.csv', index=False, encoding='utf-8-sig')
```

```python
# --- areas lost for a single locality ---
loc = '3000'

lost = unmatched[unmatched['LocalityCode'] == loc].copy()
lost['StatArea'] = pd.to_numeric(lost['StatArea'], errors='coerce')
lost = lost.sort_values('StatArea')

print(f'locality {loc} - {lost["town"].iloc[0] if len(lost) else "not in unmatched"}')
print(f'areas lost: {len(lost)}\n')
print(sorted(lost['StatArea'].dropna().astype(int).tolist()), '\n')
print(lost[['City_agas_code', 'StatArea', 'n_rows', 'first_date', 'last_date', 'total_positive']]
      .to_string(index=False))

# context: what the two sources actually contain for this locality
corona_areas = set(pd.to_numeric(
    df_corona_merged.loc[df_corona_merged['City_agas_code'].str.startswith(loc + '_'), 'City_agas_code']
    .str.split('_').str[1], errors='coerce').dropna().astype(int))
census_areas = set(pd.to_numeric(
    df_mifkad.loc[df_mifkad['City_agas_code'].str.startswith(loc + '_'), 'City_agas_code']
    .str.split('_').str[1], errors='coerce').dropna().astype(int))

print(f'\nareas in corona:  {len(corona_areas)}  range {min(corona_areas)}-{max(corona_areas)}')
print(f'areas in census:  {len(census_areas)}  range {min(census_areas)}-{max(census_areas)}')
print(f'in corona only:   {sorted(corona_areas - census_areas)}')
print(f'in census only:   {sorted(census_areas - corona_areas)}')
```
