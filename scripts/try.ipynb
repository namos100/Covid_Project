```python
# since some of the data in the Mifkad 2022 was done for combined ares in a city (City_agas_code), but have values in the corona dataset (df_loc_corona), I will retrive data from one city area and apply it ao another as well.
# for example, for city 1020 there are two areas 1 and 2 in the corona dataset, data for 1 and two was done together and was recorded in area 2 in the mifkad dataset. so I will copy the values from area 2 to area 1.



neighbor_map = {
    '1020_1': '1020_2',
    '1161_11': '1161_16',
    '229_1': '229_3',
    '2600_33': '2600_32',
    '2640_11': '2640_12',
    '3000_2112': '3000_2111',
    '3000_2211': '3000_2213',
    '3000_2212': '3000_2213',
    '3000_2311': '3000_2312',
    '3000_2412': '3000_2411',
    '3000_2613': '3000_2611',
    '3000_2812': '3000_2811',
    '3000_2911': '3000_2912',
    '4000_812': '4000_813',
    '5000_115': '5000_114',
    '5000_731': '5000_732',
    '5000_734': '5000_733',
    '6600_624': '6600_623',
    '6800_12': '6800_16',
    '7200_11': '7200_12',
    '7300_34': '7300_35',
    '7700_11': '7700_12',
    '7700_15': '7700_14',
    '7700_25': '7700_26',
    '8500_11': '8500_15',
    '8600_414': '8600_415',
    '9200_1': '9200_2',
    '9500_11': '9500_12',
}


for target_city, neighbor_city in neighbor_map.items():
    mask = merged_df.index == target_city
    neighbor_values = df_mifkad.loc[neighbor_city]
    for col in df_mifkad.columns:
        merged_df.loc[mask, col] = neighbor_values[col]

merged_df[merged_df.index == '1020_1'].head()
```

```python
from pathlib import Path
import pandas as pd

# ---------
# Config
# ---------
OUT_DIR = Path("outputs")          # change if you want
OUT_DIR.mkdir(parents=True, exist_ok=True)

# ---------
# 1) Unique keys
# ---------
loc_keys = pd.Index(df_loc_corona.index.astype(str).str.strip()).unique()
mif_keys = pd.Index(df_mifkad.index.astype(str).str.strip()).unique()

# ---------
# 2) Full-key unmatched (exact city_area, e.g., 1015_3)
# ---------
missing_in_mifkad_full = sorted(set(loc_keys) - set(mif_keys))
missing_in_loc_full    = sorted(set(mif_keys) - set(loc_keys))

df_missing_in_mifkad_full = pd.DataFrame({"city_area_key": missing_in_mifkad_full})
df_missing_in_loc_full    = pd.DataFrame({"city_area_key": missing_in_loc_full})

# Add helpful split columns for eyeballing
for df_ in (df_missing_in_mifkad_full, df_missing_in_loc_full):
    parts = df_["city_area_key"].str.split("_", n=1, expand=True)
    df_["city_base"] = parts[0]
    df_["subarea"] = parts[1].fillna("")

# Add counts per base city (so you can sort by "how bad is it")
df_missing_in_mifkad_full["n_missing_subareas_for_base"] = (
    df_missing_in_mifkad_full.groupby("city_base")["city_area_key"].transform("count")
)
df_missing_in_loc_full["n_missing_subareas_for_base"] = (
    df_missing_in_loc_full.groupby("city_base")["city_area_key"].transform("count")
)

# Save
df_missing_in_mifkad_full.sort_values(["n_missing_subareas_for_base", "city_base", "subarea"], ascending=[False, True, True]) \
    .to_csv(OUT_DIR / "missing_in_mifkad_FULLKEY.csv", index=False)

df_missing_in_loc_full.sort_values(["n_missing_subareas_for_base", "city_base", "subarea"], ascending=[False, True, True]) \
    .to_csv(OUT_DIR / "missing_in_loc_FULLKEY.csv", index=False)

print("Saved:")
print(OUT_DIR / "missing_in_mifkad_FULLKEY.csv")
print(OUT_DIR / "missing_in_loc_FULLKEY.csv")

# ---------
# (Optional but SUPER useful) 3) Base-city unmatched (e.g., 1015 only)
# ---------
loc_base = pd.Index(loc_keys.str.split("_").str[0]).unique()
mif_base = pd.Index(mif_keys.str.split("_").str[0]).unique()

missing_in_mifkad_base = sorted(set(loc_base) - set(mif_base))
missing_in_loc_base    = sorted(set(mif_base) - set(loc_base))

pd.DataFrame({"city_base": missing_in_mifkad_base}).to_csv(OUT_DIR / "missing_in_mifkad_BASE.csv", index=False)
pd.DataFrame({"city_base": missing_in_loc_base}).to_csv(OUT_DIR / "missing_in_loc_BASE.csv", index=False)

print("Also saved:")
print(OUT_DIR / "missing_in_mifkad_BASE.csv")
print(OUT_DIR / "missing_in_loc_BASE.csv")
```
